```python
# เปิด tunnel ไว้เบื้องหลัง แล้วดึงลิงก์ออกมาแสดง
import subprocess, re, time
log = open('/content/cf.log', 'w')
subprocess.Popen(['cloudflared', 'tunnel', '--url', 'http://localhost:7865'],
                 stdout=log, stderr=log)
time.sleep(8)
m = re.search(r'https://[-\w]+\.trycloudflare\.com', open('/content/cf.log').read())
print("🌐 Cloudflare URL:", m.group(0) if m else "ไม่พบ URL ลองรันใหม่")

# รัน Fooocus (ไม่ใส่ --share)
!python entry_with_update.py --port 7865 --always-high-vram
```
